```python
import random
import matplotlib.pyplot as plt
import pandas as pd
wind_effects=random.randint(0,1)
weather_type=["windy","clear","rainy"]
weather_condition=random.choice(weather_type)
print(weather_condition)
print(wind_effects)
Obstacles=random.randint(0,1)
print(Obstacles)
battery_log=100
altitude_log=0
speed_log=5
gps_x_log=0
gps_y_log=0
battery_drain=0.8
parameters=[]
for i in range(1,51):
      if(weather_condition=="windy" or weather_condition=="rainy"):
                battery_log-=(battery_drain*1.5)
                print("Battery Level: ",battery_log)
                
      else:
                battery_log-=(battery_drain)
                print("Battery Level: ",battery_log)
      if(wind_effects==1):
               speed_log=speed_log+(random.randint(-1,1))
               print("Speed :",speed_log)
      else:
              print(speed_log)
      if(Obstacles==1):
                 gps_x_log+=(random.randint(-2,2))
                 gps_y_log+=(random.randint(-2,2))
                 altitude_log+=0.5
                 print("GPS_X :",gps_x_log)
                 print("GPS_Y :",gps_y_log)
                 print("Altitude:",altitude_log)
      else:
              continue
      parameters.extend([battery_log,speed_log,gps_x_log,gps_y_log,altitude_log])
        
      for b in range(0,len(parameters),5):
        print(parameters[b:b+5])
        break

#visualisation

df = pd.read_excel("parametrs_dataset.xlsx")
df.columns = df.columns.str.strip()
if "Simulation Step" not in df.columns:
    df["Simulation Step"] = range(len(df))

print(df.columns)
fig, axes = plt.subplots(
    1, 2,
    figsize=(12, 5)
)

axes[0].plot(
    df["GPS X"],
    df["GPS Y"],
    color="gray",
    marker="o",
    markersize=5,
    linewidth=1.2,
    alpha=0.8
)

axes[0].set_title("Drone Flight Path", fontsize=11)
axes[0].set_xlabel("GPS X", fontsize=10)
axes[0].set_ylabel("GPS Y", fontsize=10)
axes[0].grid(True, alpha=0.3)
axes[1].plot(
    df["Simulation Step"],
    df["Battery %"],
    color="gray",
    linewidth=1.2,
    label="Battery %"
)

axes[1].plot(
    df["Simulation Step"],
    df["Altitude"],
    color="gray",
    linewidth=1.2,
    label="Altitude"
)

axes[1].set_title("Battery and Altitude", fontsize=11)

axes[1].set_xlabel("Simulation Step", fontsize=10)
axes[1].set_ylabel("Value", fontsize=10)

axes[1].grid(True, alpha=0.3)

axes[1].legend(
    loc="upper right",
    fontsize=8
)
plt.tight_layout()
plt.show()
```
